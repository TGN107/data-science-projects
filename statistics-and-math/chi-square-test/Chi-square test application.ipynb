```python

# Step 2: Read the uploaded CSV file (replace with your actual filename if needed)
df = pd.read_csv('StudentsPerformance.csv')

# Step 3: Preview the dataset
df.head()
```

```python
cat1 = 'parental level of education'
cat2 = 'test preparation course'

# Get unique values
cat1_values = df[cat1].unique()
cat2_values = df[cat2].unique()

print("Unique values of Parental Level of Education:", cat1_values)
print("Unique values of Test Preparation Course:", cat2_values)
```

Count Observed Frequencies (O)

We will construct an observed frequency table using nested loops and value counting.




```python
# Build observed frequency table manually
observed = []

for level in cat1_values:
    row = []
    for prep in cat2_values:
        count = len(df[(df[cat1] == level) & (df[cat2] == prep)])
        row.append(count)
    observed.append(row)
```

```python
# Convert to DataFrame for easy visualization
observed = np.array(observed)
print("\nObserved Frequency Table (O):")
print(pd.DataFrame(observed, index=cat1_values, columns=cat2_values))
```

Compute Totals

```python
# Row Totals, Column Totals, Grand Total
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)
grand_total = observed.sum()

print("\nRow Totals:", row_totals)
print("Column Totals:", col_totals)
print("Grand Total:", grand_total)
```

```python
expected = np.zeros_like(observed, dtype=float)

for i in range(len(row_totals)):
    for j in range(len(col_totals)):
        expected[i][j] = (row_totals[i] * col_totals[j]) / grand_total

print("\nExpected Frequency Table (E):")
print(pd.DataFrame(expected, index=cat1_values, columns=cat2_values))
```

```python
chi_square_stat = 0

for i in range(len(row_totals)):
    for j in range(len(col_totals)):
        chi_square_stat += ((observed[i][j] - expected[i][j]) ** 2) / expected[i][j]

print(f"\nChi-Square Statistic (X²): {chi_square_stat:.4f}")
```

```python
r = len(row_totals)
c = len(col_totals)
df = (r - 1) * (c - 1)

print(f"Degrees of Freedom: {df}")
```

Get P-Value (Optional: using scipy just for result)

```python
from scipy.stats import chi2

p_value = 1 - chi2.cdf(chi_square_stat, df)
print(f"P-Value: {p_value:.4f}")
```

Final Interpretation

```python
alpha = 0.05
if p_value < alpha:
    print("\n Reject Null Hypothesis: Variables are dependent (associated).")
else:
    print("\n Fail to Reject Null Hypothesis: Variables are independent.")
```
