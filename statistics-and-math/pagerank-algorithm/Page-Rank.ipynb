 Imagine this: You’re voting for the most popular website.
If a website (say Page A) is linked to by many other websites, then it's probably important, right?

And if those websites themselves are important, their links are worth even more.

That’s exactly what PageRank does: it measures how important a web page is based on who links to it.
Absolutely! Here's a **line-by-line explanation** of your PageRank implementation:

First, Understand the Direction:
In PageRank, the link matrix is column-based:

Each column represents a source page (the page giving the link),

Each row represents a target page (the page receiving the link).

---
For example:
### ✅ **Step 1: Build Link Matrix**

```python
import numpy as np
```

* Import the NumPy library for numerical operations, especially with arrays.

```python
# Pages:  A = 0, B = 1, C = 2
link_matrix = np.array([
    [0, 0, 1],  # A gets link from C
    [1, 0, 1],  # B gets link from A and C
    [0, 1, 0]   # C gets link from B
], dtype=float)
```

* This is your **adjacency matrix** of the graph.
* Each **column** represents a page giving links **to others**.

  * `link_matrix[i][j] = 1` means **Page j links to Page i**.
* In this matrix:

  * **Column 0 (A)**: A links to B → only `B (row 1)` gets `1`
  * **Column 1 (B)**: B links to C → only `C (row 2)` gets `1`
  * **Column 2 (C)**: C links to A and B → A (row 0) and B (row 1) get `1`

---

### ✅ **Step 2: Convert to Stochastic Matrix**

```python
n = link_matrix.shape[0]  # number of pages
```

* `n = 3`, number of nodes/pages.

```python
for j in range(n):
    col_sum = link_matrix[:, j].sum()
    if col_sum != 0:
        link_matrix[:, j] /= col_sum
    else:
        link_matrix[:, j] = 1.0 / n
```

* Loop through **each column (each page’s outlinks)**.
* Normalize the column to sum to 1 so that it becomes a **probability distribution**:

  * For example, if Page C links to A and B, its column will have two `1`s → sum = 2 → divide by 2.
* If a page has **no outlinks (dangling node)**, assume it links equally to all pages.

---

### ✅ **Step 3: PageRank Formula**

```python
damping = 0.85
```

* Damping factor: probability of following a link (85%). The rest (15%) is **random jump** to any page.

```python
pagerank = np.ones(n) / n
```

* Initialize equal PageRank for all pages, i.e., `[1/3, 1/3, 1/3]`.

```python
tolerance = 1e-6
```

* Convergence threshold: when the PageRank updates change less than this, stop iterating.

```python
for _ in range(100):
    new_rank = (1 - damping) / n + damping * link_matrix @ pagerank
    if np.linalg.norm(new_rank - pagerank, 1) < tolerance:
        break
    pagerank = new_rank
```

* Iterate up to 100 times or until convergence.
* Formula:

  $$
  \text{new\_rank} = \frac{1 - d}{N} + d \cdot (M \cdot \text{pagerank})
  $$

  where:

  * $d$ = damping factor
  * $M$ = stochastic matrix (your `link_matrix`)
  * $\text{pagerank}$ = vector of previous ranks
* `@` is matrix multiplication.
* Use **L1 norm** (`np.linalg.norm(..., 1)`) to check convergence.

---

### ✅ **Step 4: Show the Result**

```python
for i, score in enumerate(pagerank):
    print(f"Page {chr(65+i)}: {score:.4f}")
```

* Iterate over final `pagerank` values.
* Convert index `0 → A`, `1 → B`, `2 → C` using `chr(65+i)` (ASCII A = 65).
* Print the score rounded to 4 decimal places.

---

### 🔚 Final Output Example

```txt
Page A: 0.3871
Page B: 0.4435
Page C: 0.1694
```

This shows the **relative importance** of pages in the network — more backlinks and from more important pages = higher rank.

````python
import numpy as np
````

````python
link_matrix = np.array([
    [0, 0, 0, 0],  # A
    [1, 0, 0, 1],  # B gets from A
    [0, 1, 1, 0],  # C gets from B
    [0, 0, 1, 0]   # D gets from C
],dtype=np.float64)
````

````python
n = link_matrix.shape[0]  # number of pages
````

````python
for j in range(n):
    col_sum = link_matrix[:, j].sum()
    if col_sum != 0:
        link_matrix[:, j] /= col_sum
    else:
        link_matrix[:, j] = 1.0 / n
````

````python
damping = 0.85
pagerank = np.ones(n) / n
tolerance= 1e-6
````

````python
for _ in range(100):
    new_rank = (1 - damping) / n + damping * link_matrix @ pagerank
    if np.linalg.norm(new_rank - pagerank, 1) < tolerance:
        break
    pagerank = new_rank
````

````python
for i, score in enumerate(pagerank):
    print(f"Page {chr(65+i)}: {score:.4f}")
````
